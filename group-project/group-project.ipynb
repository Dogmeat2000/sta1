| Item                                                                                               | Your notes                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                  |
|----------------------------------------------------------------------------------------------------|-----------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------|
| Engineering objective and intended decision                                                        | Give buyers a guideline for choosing an efficient 2026 EV: which wheel size, drive system and battery size go with the lowest highway energy use, checked with like-for-like comparisons of the same car.                                                                                                                                                                                                                                                                                                                                                                                                                                                                   |
| Target population and observational unit                                                           | All model-year 2026 battery-electric configurations certified for sale in the US. Unit: one EPA-label configuration; n = 333.                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                               |
| Dataset source and study design                                                                    | Data Source is the Electric Vehicles data from the US Department of Energy for 2026.<br/>Study design is described in the attached Guide (FEG2026.pdf).<br/>The dataset was sourced from https://www.fueleconomy.gov/feg/download.shtml<br/>Observational near-census of lab-tested label values: supports association, not causation. Variants of one platform are not independent.                                                                                                                                                                                                                                                                                        |
| Primary statistical question                                                                       | For the same car, how much does highway energy consumption (kWh/100 mi) change between its smallest and largest wheel option?                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                               |
| Secondary statistical questions                                                                    | 1. Among cars and SUVs, does mean highway consumption differ between drive systems, and does the difference hold within the same model?<br/>2. Does a larger battery go with higher highway consumption, and which configurations beat their prediction?                                                                                                                                                                                                                                                                                                                                                                                                                    |
| Method 1: session/chapter, parameter, variables, why this method, assumptions, diagnostics, output | Session 8 / Ch. 8, paired comparison. Parameter: mean within-car difference μ_d (largest − smallest wheels). Variables: highway consumption; wheel size from the vehicle name; 44 cars sold with several wheel sizes, otherwise identical. Why: each car is compared with itself, so vehicle size and brand cannot confound the wheel effect. Assumptions: independent cars, approximately normal differences. Diagnostics: table of pairs, QQ-plot of differences. Output: mean difference with 95% CI, paired t, p.                                                                                                                                                       |
| Method 2: session/chapter, parameter, variables, why this method, assumptions, diagnostics, output | Session 9 / Ch. 9, one-way ANOVA. Parameters: mean highway consumption for FWD (21), RWD (50), AWD (198) and 4WD (27) cars and SUVs; pickups excluded, as 34 of 37 are 4WD. Why: drive system is a choice buyers make. Assumptions: independent groups, similar variances, normal residuals. Diagnostics: boxplots, group SDs, QQ-plot; Welch with Games–Howell if variances differ, otherwise Tukey. Output: F, df, p, pairwise differences with CIs, η².<br/>Planned supplementary check: same-model 2WD versus AWD variants on the same battery (at least 20 models), paired differences, to test whether the across-car difference survives a like-for-like comparison. |
| Method 3: session/chapter, parameter, variables, why this method, assumptions, diagnostics, output | Session 10 / Ch. 10, simple linear regression. Parameter: slope β₁, change in mean highway consumption per kWh of battery energy (voltage × Ah / 1000). Why: tests whether buying range costs efficiency. Assumptions: linear mean, independent errors, constant variance, normal errors. Diagnostics: residual-versus-fitted, QQ-plot, leverage, Cook's distance. Output: β̂₁ with 95% CI, t, p, s, R², and the configurations outside the 95% prediction interval.                                                                                                                                                                                                        |
| Practical threshold or decision criterion                                                          | A difference of at least 2 kWh/100 mi (about 5% of typical use) is practically relevant, for wheel options, drive systems and a 50 kWh change in battery size. Two-sided tests at α = 0.05.                                                                                                                                                                                                                                                                                                                                                                                                                                                                                 |
| Missing-data and unusual-observation plan                                                          | Carry each title row's vehicle name down, drop title rows, keep one row per configuration, parse decimal commas. Key variables have no missing values; wheel size is named for 147 configurations. Influential or outlying cases are investigated and reported with and without them, never deleted for being inconvenient.                                                                                                                                                                                                                                                                                                                                                 |
| Planned versus exploratory analyses                                                                | Planned: Methods 1–3, the same-model drive check and the threshold, fixed before inference. Exploratory, labelled as such: body type, brand, highway range, battery chemistry (only 4 LFP); any pattern found here is a hypothesis, not a conclusion.                                                                                                                                                                                                                                                                                                                                                                                                                       |
| Integrated reporting structure                                                                     | Objective → data audit → descriptive plots → Methods 1–3 → guideline per body type: which wheel size, drive system and battery size to prefer, with effect sizes, and where the across-car and same-car comparisons disagree → limitations → AI use.                                                                                                                                                                                                                                                                                                                                                                                                                        |
